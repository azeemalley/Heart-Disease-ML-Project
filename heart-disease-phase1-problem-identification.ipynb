## 4. Methodology

### 4.1 Approach Overview

We will implement a **5-stage pipeline**:

```
Multi-Center Data → EDA & Visualization → Feature Engineering → 
Model Building & Stacking → Explainability & Comparison
```

### 4.2 Detailed Methodology

**Stage 1 — Data Collection & Fusion (Phase 2):**
- Use the combined UCI dataset (Cleveland + Hungary + Switzerland + VA) — 920 records
- This is larger and more diverse than the commonly-used Cleveland-only (303 records)
- Handle missing values with median imputation for numeric, mode for categorical

**Stage 2 — Exploratory Data Analysis (Phase 2):**
- Statistical analysis: describe(), missing values, skewness, kurtosis
- Visual analysis: distributions, correlation heatmaps, box plots by target, violin plots
- Identify outliers using IQR method

**Stage 3 — Feature Engineering (Phase 3):**
- Run 5 feature importance algorithms: Random Forest, XGBoost, LightGBM, Permutation Importance, SHAP
- Create 8+ clinically-informed new features:
  - `age_hr_ratio` = age / thalach (cardiovascular efficiency)
  - `bp_age_risk` = trestbps × age (cumulative blood pressure load)
  - `chol_age_index` = chol / age (age-adjusted cholesterol)
  - `hr_reserve` = 220 - age - thalach (heart rate reserve)
  - `exercise_risk_score` = oldpeak × exang (combined exercise risk)
  - `metabolic_score` = chol + trestbps + (fbs × 50)
  - `vessel_thal_interaction` = ca × thal (structural cardiac risk)
  - `chest_pain_gender` = cp × sex (gender-pain interaction)
- Apply K-Means clustering to create a `cardiac_cluster` feature
- Standardization using RobustScaler (handles outliers better than StandardScaler)

**Stage 4 — Model Building (Phase 4):**
- Train 5 base models: Logistic Regression, Random Forest, XGBoost, LightGBM, SVM
- Build a **Stacking Ensemble** with meta-learner (Logistic Regression)
- Hyperparameter tuning with Optuna (Bayesian optimization)
- Evaluation: Accuracy, F1-Score, ROC-AUC, Precision-Recall AUC, Confusion Matrix
- 10-fold Stratified Cross-Validation

**Stage 5 — Explainability & Research Comparison (Phase 5):**
- SHAP waterfall plots, beeswarm plots, force plots
- LIME local explanations for individual predictions
- Compare with 5 recent research papers (2022–2024)
- Identify and justify the novelty contribution

## 5. Novelty Contribution

Our work makes the following **unique contributions** that distinguish it from existing research:

### 🔬 Contribution 1: Multi-Center Dataset Fusion
Most existing papers use **only the Cleveland dataset (303 samples)**. We use the **full 4-center UCI dataset (920 samples)** — Cleveland, Hungary, Switzerland, and VA Long Beach — significantly improving model generalizability.

### 🧠 Contribution 2: Clinically-Informed Feature Engineering
We create **8 novel interaction features derived from medical domain knowledge** (e.g., `age_hr_ratio`, `bp_age_risk`, `hr_reserve`). These are not simple polynomial transformations — they encode clinical relationships validated in cardiology literature. No existing paper in our comparison has done this systematically.

### 🏗️ Contribution 3: Stacking Ensemble with Heterogeneous Base Learners
We use a **stacking meta-ensemble** combining Logistic Regression + Random Forest + XGBoost + LightGBM + SVM. Most papers compare algorithms independently without stacking.

### 🔍 Contribution 4: Full Explainability Pipeline
We implement both **SHAP (global)** and **LIME (local)** explanations — providing both global model understanding and individual patient-level explanations. This addresses the clinical adoption barrier (black-box concern).

### ⚖️ Contribution 5: Handling Class Imbalance with SMOTE+Tomek
We apply combined **SMOTE + Tomek Links** resampling — more sophisticated than simple oversampling used in most existing works.
